## Slicing a list

A slice uses `values[start:stop:step]`. The stop position is excluded, and slicing creates a new list.

````python
numbers = [10, 20, 30, 40, 50, 60, 70]

print(numbers[1:4])
print(numbers[:3])
print(numbers[4:])
print(numbers[::2])
print(numbers[::-1])
print(numbers[-3:])
````

## List operators and membership

````python
group_a = ["Anu", "Ravi"]
group_b = ["Sara", "Imran"]

print(group_a + group_b)
print([0] * 5)
print("Ravi" in group_a)
print("ravi" in group_a)
print([1, 2] == [1, 2])
print([1, 2] == [2, 1])
````

# Stage 2 — Lists are mutable

Unlike strings, lists can be changed after creation. You can replace one element or an entire slice.

````python
scores = [60, 70, 80, 90]
print("Original:", scores)

scores[0] = 65
print("After one replacement:", scores)

scores[1:3] = [72, 82]
print("After slice replacement:", scores)

scores[2:2] = [75, 77]
print("After inserting through a slice:", scores)
````

## Adding elements: `append()`, `extend()`, and `insert()`

- `append(value)` adds one value to the end.
- `extend(iterable)` adds each value from another iterable.
- `insert(index, value)` adds one value at a selected position.

````python
shopping = ["rice", "milk"]

shopping.append("bread")
print("After append:", shopping)

shopping.extend(["eggs", "fruit"])
print("After extend:", shopping)

shopping.insert(1, "tea")
print("After insert:", shopping)
````

Predict the two final lists before running this cell. Explain why they are different.

## Removing elements: `remove()`, `pop()`, `del`, and `clear()`

- `remove(value)` deletes the first matching value.
- `pop(index)` removes and returns an element; the default is the last element.
- `del` removes by index or slice.
- `clear()` removes all elements.

````python
tasks = ["read", "code", "test", "code", "submit"]

tasks.remove("code")
print("After remove:", tasks)

completed = tasks.pop(0)
print("Popped value:", completed)
print("After pop:", tasks)

del tasks[1:3]
print("After del slice:", tasks)

tasks.clear()
print("After clear:", tasks)
````

## Finding and counting: `index()` and `count()`

`index(value)` returns the first position but raises an error if the value is missing. Check membership before using it when absence is possible.

````python
ratings = [4, 5, 3, 5, 4, 5]

print("Number of fives:", ratings.count(5))
print("First five at index:", ratings.index(5))

target = 2
if target in ratings:
    print("First occurrence at:", ratings.index(target))
else:
    print(target, "is not present")
````

# Stage 3 — Traversing lists and building algorithms

A loop lets us inspect every element. Many list algorithms use a counter, total, flag, or best-so-far value.

````python
temperatures = [29.5, 31.0, 30.2, 32.4, 28.9]
total = 0
hot_days = 0

for temperature in temperatures:
    total += temperature
    if temperature > 30:
        hot_days += 1

average = total / len(temperatures)
print("Average:", round(average, 2))
print("Days above 30°C:", hot_days)
````

## Values versus indices

Loop directly over values when positions are unnecessary. Use `range(len(list_name))` when you need indices. `enumerate()` conveniently supplies both.

````python
subjects = ["Python", "Statistics", "Database"]

print("Values only:")
for subject in subjects:
    print(subject)

print("\nIndex and value:")
for index, subject in enumerate(subjects):
    print(index, subject)

print("\nNumbered from one:")
for number, subject in enumerate(subjects, start=1):
    print(number, subject)
````

## Built-in functions with numerical lists

`len()`, `sum()`, `min()`, and `max()` solve common tasks. You should also understand how to reproduce their logic with loops.

````python
marks = [78, 64, 91, 83, 55]

print("Count:", len(marks))
print("Total:", sum(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Average:", round(sum(marks) / len(marks), 2))
````

# Stage 4 — Sorting, reversing, copying, and transformation

## `sort()` versus `sorted()`

`list.sort()` changes the list and returns `None`. `sorted(iterable)` returns a new sorted list and preserves the original.

````python
values = [42, 7, 19, 7, 31]
ascending_copy = sorted(values)
descending_copy = sorted(values, reverse=True)

print("Original:", values)
print("Ascending copy:", ascending_copy)
print("Descending copy:", descending_copy)

sort_result = values.sort()
print("After values.sort():", values)
print("Return value of sort():", sort_result)
````

## Sorting text

Default text sorting is case-sensitive. A key can describe how values should be compared.

````python
names = ["zara", "Anu", "ravi", "Bina"]

print("Default:", sorted(names))
print("Ignore case:", sorted(names, key=str.lower))
print("By length:", sorted(names, key=len))
````

## Aliasing and copying

Assignment does not copy a list. Two names may refer to the same list. Use `copy()` or a full slice when an independent shallow copy is needed.

````python
original = [10, 20, 30]
alias = original
copied = original.copy()
sliced_copy = original[:]

alias.append(40)
copied.append(50)
sliced_copy.append(60)

print("Original:", original)
print("Alias:", alias)
print("Copied:", copied)
print("Slice copy:", sliced_copy)
print("original is alias:", original is alias)
print("original is copied:", original is copied)
````

## Reverse carefully

Compare `reverse()`, `reversed()`, and `[::-1]`. Predict which ones change the source list.

````python
letters = ["a", "b", "c", "d"]
slice_reversed = letters[::-1]
built_in_reversed = list(reversed(letters))

print("Before reverse():", letters)
letters.reverse()
print("After reverse():", letters)
print("Slice result:", slice_reversed)
print("reversed() result:", built_in_reversed)
````

## Optional preview: list comprehensions

A list comprehension is a compact way to build a list. First write the loop version; then compare it with the comprehension.

````python
numbers = [1, 2, 3, 4, 5, 6]

squares_loop = []
for number in numbers:
    squares_loop.append(number ** 2)

squares_comprehension = [number ** 2 for number in numbers]
even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print(squares_loop)
print(squares_comprehension)
print(even_squares)
````

# Stage 5 — Nested lists: small tables

A nested list contains other lists. It can represent rows and columns of a small table.

````python
students = [
    ["Anu", 78, 84, 81],
    ["Ravi", 65, 72, 70],
    ["Sara", 91, 88, 94]
]

print("First row:", students[0])
print("First name:", students[0][0])
print("Sara's second mark:", students[2][2])

for row in students:
    name = row[0]
    marks = row[1:]
    average = sum(marks) / len(marks)
    print(name, "average:", round(average, 2))
````

## Matrix traversal

Use one loop for rows and another for the values inside each row.

````python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [7, 8, 9]
]

grand_total = 0
for row in matrix:
    row_total = 0
    for value in row:
        row_total += value
    grand_total += row_total
    print("Row total:", row_total)

print("Grand total:", grand_total)
````
