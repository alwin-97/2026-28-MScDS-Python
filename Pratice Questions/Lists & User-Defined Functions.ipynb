# Stage 2 — Define, call, and return · 35 minutes

A **parameter** is a name in a function definition. An **argument** is a value supplied in a call.

In `def double(value):`, `value` is a parameter. In `double(6)`, `6` is an argument. Defining a function stores its behaviour; calling it runs its body.

### Four common input/output patterns

These are ways of designing functions, not four separate Python data types.

| Pattern | Example purpose |
|---|---|
| No parameters, no explicit return | Print a fixed heading |
| Parameters, no explicit return | Display supplied data |
| No parameters, returned value | Create a fresh sample list |
| Parameters, returned value | Calculate a result from supplied data |

Every Python function returns something. Reaching the end without `return`, or using a bare `return`, returns `None`.

````python
def show_heading():
    print("Reading summary")

def display_count(values):
    print("Number of values:", len(values))

def sample_values():
    return [2, 4, 6]

def double(value):
    return value * 2

show_heading()
display_count([10, 20])
data = sample_values()
print("Sample:", data)
print("Doubled:", double(6))
````

# Stage 4 — Variable-length and keyword-only parameters · 45 minutes

### `*args`: accept a variable number of positional arguments

The star collects extra positional arguments into a **tuple**. The name `args` is conventional, not compulsory. `*values` and `*lists` are equally valid.

In a function **call**, a star does the reverse: it unpacks a sequence into separate positional arguments.

````python
def combine_totals(*values):
    return sum(values)

print(combine_totals())
print(combine_totals(4, 8, 12))
measurements = [4, 8, 12]
print(combine_totals(*measurements))
# combine_totals(measurements) supplies ONE argument: a list.
# It does not supply three separate numbers.
````

### `**kwargs`: accept named arguments

A double star collects extra keyword arguments into a dictionary. In a call, `**options` unpacks a dictionary into named arguments. Names must be strings and must match the receiving function unless it accepts arbitrary keywords.

````python
def describe_session(**details):
    for name, value in details.items():
        print(name, "=", value)

describe_session(topic="Lists", duration=60)

def shift_reading(value, offset=0):
    return value + offset

options = {"value": 10, "offset": 3}
print(shift_reading(**options))
````

### Keyword-only parameters

A parameter after a bare `*` or after `*args` must be passed by name. This makes calls such as `round_values(data, digits=2)` explicit.

````python
def round_values(values, *, digits=1):
    return [round(value, digits) for value in values]

print(round_values([1.234, 5.678], digits=2))
# round_values([1.234], 2)  # TypeError: digits must be named.

def weighted_total(*values, multiplier=1):
    return sum(values) * multiplier

print(weighted_total(2, 3, multiplier=10))
````

### Positional-only parameters: recognise the slash

Parameters before `/` must be passed by position. Parameters between `/` and `*` can be positional or keyword; parameters after `*` must be keyword arguments.

````python
def repeat_item(item, /, times=2, *, uppercase=False):
    if uppercase:
        item = item.upper()
    return [item] * times

print(repeat_item("lab", times=3, uppercase=True))
# repeat_item(item="lab")  # TypeError: item is positional-only.
````

# Stage 5 — List algorithms and function behaviour · 45 minutes

### Mutating an argument is different from rebinding a local name

Function parameters are local names referring to the supplied objects. Changing a shared list changes the caller's list. Reassigning the local parameter to a new list does not reassign the caller's variable.

````python
def append_in_place(values):
    values.append(99)

def rebind_locally(values):
    values = [0, 0]
    return values

source = [1, 2]
append_in_place(source)
print("After mutation:", source)
replacement = rebind_locally(source)
print("Caller still has:", source)
print("Returned replacement:", replacement)
````

### Mutable default arguments

Default values are evaluated when the function is defined. A default list can therefore be shared across calls. Use `None` and create a fresh list inside the function when a fresh container is intended.

````python
def unsafe_collect(value, collected=[]):
    collected.append(value)
    return collected

print(unsafe_collect("first"))
print(unsafe_collect("second"))  # Includes the earlier value.

def collect(value, collected=None):
    if collected is None:
        collected = []
    collected.append(value)
    return collected

print(collect("first"))
print(collect("second"))
# If a list is explicitly supplied to collect(), that list is mutated.
````

### Local scope

A variable created inside a function is local to that call. Return the result to use it outside. Avoid global variables for calculations in this workbook.

````python
total = 100

def calculate_total(values):
    total = sum(values)
    return total

answer = calculate_total([1, 2, 3])
print("Returned total:", answer)
print("Outer total:", total)
````

### Recursive functions

A recursive function calls itself on a smaller problem. It needs a **base case** and progress toward it. Use small inputs here; deep recursion can reach Python's recursion limit.

````python
def recursive_total(values):
    if not values:
        return 0
    return values[0] + recursive_total(values[1:])

print(recursive_total([2, 4, 6]))
print(recursive_total([]))
````

### Generator functions: yield one value at a time

A function containing `yield` produces a generator when called. Iteration starts or resumes its body. Unlike a returned list, a generator is consumed as you iterate.

````python
def generate_positive(values):
    for value in values:
        if value > 0:
            yield value

stream = generate_positive([-1, 2, 0, 4])
print(list(stream))
print(list(stream))  # Already exhausted.
print(list(generate_positive([-1, 2, 0, 4])))  # A fresh generator.
````

## Quick reference

| Concept | Pattern | Meaning |
|---|---|---|
| Definition | `def calculate(values):` | Creates a named function |
| Positional argument | `calculate([1, 2])` | Matches by position |
| Keyword argument | `calculate(values=[1, 2])` | Matches by parameter name |
| Default | `def f(values, limit=10):` | Uses 10 when omitted |
| Returned value | `return result` | Ends the call and sends data back |
| Multiple values | `return total, average` | Returns a tuple |
| Variable positional arguments | `def f(*values):` | Collects positional arguments in a tuple |
| Variable keyword arguments | `def f(**options):` | Collects keywords in a dictionary |
| Keyword-only | `def f(values, *, digits=2):` | digits must be named |
| Positional-only | `def f(value, /):` | value cannot be supplied by name |
| Unpack a sequence in a call | `f(*values)` | Supplies separate positional arguments |
| Unpack a dictionary in a call | `f(**options)` | Supplies named arguments |
| Safe fresh mutable default | `def f(values=None):` | Create a list inside if values is None |
| Lambda | `lambda x: x * 2` | One-expression function |
| Generator | `yield value` | Produces values during iteration |
| Recursion | `f(smaller_problem)` | Calls itself; needs a base case |
