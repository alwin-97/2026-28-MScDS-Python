## A1. Variables, constants, and built-in functions

A variable is a name associated with a value. Python determines the data type from the assigned value.

````python
student_name = "Anu"
age = 21
average_mark = 78.456
is_present = True

print(student_name, type(student_name))
print(age, type(age))
print(round(average_mark, 2), type(average_mark))
print(is_present, type(is_present))
````

Python does not technically prevent a constant from changing. By convention, a name written in uppercase tells programmers that its value should remain unchanged.

````python
PASS_MARK = 50
marks = [72, 65, 81, 49]

print("Number of marks:", len(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Total:", sum(marks))
print("Average:", round(sum(marks) / len(marks), 2))
print("Required pass mark:", PASS_MARK)
````

### Input and type conversion

`input()` always returns text. Convert it before performing numerical calculations.

````python
quantity = int(input("Enter quantity: "))
unit_price = float(input("Enter unit price: "))
total = quantity * unit_price

print("Total amount: ₹", round(total, 2), sep="")
````

## A2. Branching and looping

````python
mark = 68

if mark < 0 or mark > 100:
    print("Invalid mark")
elif mark >= 75:
    print("Distinction")
elif mark >= 50:
    print("Pass")
else:
    print("Needs improvement")
````

````python
print("Using a for loop:")
for number in range(1, 6):
    print(number, end=" " )

print("\n\nUsing a while loop:")
count = 1
while count <= 5:
    print(count, end=" " )
    count += 1
````

> **Checkpoint:** Use a `for` loop when iterating through a known sequence or range. A `while` loop is useful when repetition should continue until a condition changes. Always ensure that a `while` loop can eventually stop.
