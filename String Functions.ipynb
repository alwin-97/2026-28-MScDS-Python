# Stage 2 — Indexing, slicing, operators, and immutability

A slice follows `text[start:stop:step]`. The `stop` position is excluded.

````python
word = "PYTHON"

print(word[0:3])       # positions 0, 1, 2
print(word[:4])        # from the beginning
print(word[2:])        # to the end
print(word[::2])       # every second character
print(word[::-1])      # reverse
print(word[1:5:2])
````

## String operators

Strings support concatenation (`+`), repetition (`*`), membership (`in`, `not in`), and comparisons. Comparisons are case-sensitive.

````python
first = "data"
second = "science"

print(first + " " + second)
print("ha" * 3)
print("sci" in second)
print("Data" == "data")
print("apple" < "banana")
````

# Stage 3 — String method laboratory

A method is called using dot notation: `text.method()`. Most string methods return a new value; they do not change the original string.

````python
sample = "  pYtHoN for DATA science  "

print("Original :", repr(sample))
print("lower    :", sample.lower())
print("upper    :", sample.upper())
print("title    :", sample.title())
print("capitalize:", sample.capitalize())
print("swapcase :", sample.swapcase())
print("strip    :", repr(sample.strip()))
print("Original again:", repr(sample))
````

````python
sentence = "Python is readable. Python is productive."

print(sentence.count("Python"))
print(sentence.find("readable"))
print(sentence.find("Java"))      # -1 means not found
print(sentence.startswith("Python"))
print(sentence.endswith("."))
print(sentence.replace("Python", "Programming", 1))
````

### `find()` and `index()` are similar—but not identical

Both locate a substring. `find()` returns `-1` when it is absent; `index()` raises an error. Test both and decide which is safer when absence is expected.

````python
text = "data science"
print(text.find("science"))
print(text.index("science"))

# Try a missing substring with find(), then with index().
# print(text.find("Python"))
# print(text.index("Python"))
````

## Asking questions about a string

The methods below return `True` or `False`. Test values containing spaces, signs, decimal points, or mixed characters.

````python
test_values = ["Python", "MDS501", "501", "data science", "   ", "3.14", "-25"]

for value in test_values:
    print(repr(value))
    print("  isalpha :", value.isalpha())
    print("  isdigit :", value.isdigit())
    print("  isalnum :", value.isalnum())
    print("  isspace :", value.isspace())
````

# Stage 6 — A little bit of lists: `split()` and `join()`

`split()` breaks a string into a list of strings. `join()` combines strings from a list using a separator.

````python
sentence = "Python makes text processing approachable"
words = sentence.split()

print(words)
print(type(words))
print("First word:", words[0])
print("Number of words:", len(words))
print("Rejoined:", " | ".join(words))
````

````python
record = "MDS001,Anu Thomas,Data Science"
fields = record.split(",")

for field in fields:
    print(field.strip())

clean_record = ";".join(fields)
print(clean_record)
````
