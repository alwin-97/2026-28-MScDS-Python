## A2. Anatomy of a user-defined function

```python
def function_name(parameter):
    """Briefly describe the function."""
    result = ...
    return result
```

- A **parameter** is a name in the function definition.
- An **argument** is the value supplied in a function call.
- The function body runs only when the function is called.
- `return` sends a result back to the caller and ends that function call.

````python
def calculate_rectangle_area(length, width):
    """Return the area of a rectangle."""
    area = length * width
    return area

room_area = calculate_rectangle_area(8, 5)
print("Room area:", room_area)
print("Double the area:", room_area * 2)
````

## Printing is not returning

Predict the value stored in `result_a` and `result_b`, then run the cell. A function with no explicit `return` returns `None`.

````python
def display_square(number):
    print(number ** 2)

def calculate_square(number):
    return number ** 2

result_a = display_square(6)
result_b = calculate_square(6)

print("result_a:", result_a)
print("result_b:", result_b)
````

> **Design rule for this lab:** calculation functions should usually return data. The main program decides how to display that data. This makes a function easier to reuse and test.

## A3. Four basic forms of functions

| Form | Example purpose | Reusability |
|---|---|---|
| No parameters, no return value | Display a fixed menu | Limited |
| Parameters, no return value | Display a supplied heading | Moderate |
| No parameters, return value | Obtain a value from one fixed source | Moderate |
| Parameters and return value | Calculate a result from supplied data | High |

For most problem-solving functions, parameters and return values are preferred.

````python
def display_menu():
    print("1. Add")
    print("2. Exit")

def display_heading(heading):
    print("===", heading, "===")

def get_course_code():
    return "MDS501-1"

def calculate_percentage(score, maximum):
    return score / maximum * 100

display_heading(get_course_code())
display_menu()
percentage = calculate_percentage(42, 50)
print("Percentage:", round(percentage, 1))
````

## A4. Positional, keyword, and default arguments

Arguments can be matched by position or parameter name. A default value makes an argument optional. Required parameters must come before default parameters.

````python
def calculate_simple_interest(principal, rate=6.5, years=1):
    return principal * rate * years / 100

print(calculate_simple_interest(10000))
print(calculate_simple_interest(10000, 7.0, 2))
print(calculate_simple_interest(10000, years=3))
print(calculate_simple_interest(rate=8.0, years=2, principal=10000))
````

## A5. Returning multiple values

Python can return several values. The caller can unpack them into separate variables.

````python
def summarize(values):
    """Return the minimum, maximum, and average of a non-empty list."""
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return minimum, maximum, average

low, high, mean = summarize([68, 75, 81, 92])
print("Minimum:", low)
print("Maximum:", high)
print("Average:", round(mean, 2))
````

## A6. Variable-length positional arguments

A parameter beginning with `*` collects extra positional arguments into a tuple. Use it only when the number of supplied values may genuinely vary.

````python
def calculate_total(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(calculate_total(5, 10))
print(calculate_total(5, 10, 15, 20))
print(calculate_total())
````

## A7. Local scope

A variable created inside a function is normally local to that call. Prefer returning a result over modifying a global variable.

````python
tax_rate = 0.05

def calculate_tax(amount, rate):
    tax = amount * rate
    return tax

tax_amount = calculate_tax(1200, tax_rate)
print("Tax:", tax_amount)
# print(tax)  # Uncomment to see why a local variable is unavailable here.
````
