## Character tests

Methods such as `isalpha()`, `isdigit()`, `isalnum()`, `isspace()`, `islower()`, and `isupper()` return Boolean values. A loop can apply them one character at a time.

````python
sample = "MDS 501!"

for character in sample:
    if character.isalpha():
        category = "letter"
    elif character.isdigit():
        category = "digit"
    elif character.isspace():
        category = "space"
    else:
        category = "other"
    print(repr(character), "->", category)
````
