# Stage 1 — Understand mappings · 15 minutes

A **dictionary** maps a **key** to a **value**. Use a key to find associated information. A list usually answers “what is at this position?”; a dictionary answers “what belongs to this key?”

| Structure | Example | Access |
|---|---|---|
| List | `["Anu", "Ravi"]` | `names[0]` |
| Dictionary | `{"MDS001": "Anu", "MDS002": "Ravi"}` | `students["MDS001"]` |

A dictionary is mutable. Keys are unique; values can repeat. Values may be strings, numbers, lists, or other dictionaries.

````python
student = {"register": "MDS001", "name": "Anu", "mark": 78}
print(student)
print(type(student))
print("Number of key-value pairs:", len(student))
print("Name:", student["name"])
print("Mark:", student["mark"])
````

### Square brackets versus `get()`

Use `d[key]` when the key should exist. A missing key raises `KeyError`.
Use `d.get(key, default)` when absence is expected. Without a default, `get()` returns `None` for a missing key. It does **not** insert the key.

````python
profile = {"name": "Anu", "year": 2026}
print(profile["name"])
print(profile.get("email"))
print(profile.get("email", "Not provided"))
print("email" in profile)
print(profile)
````

# Stage 3 — Update and remove · 35 minutes

Assignment adds a new key or replaces an existing value. `update()` applies several pairs at once. It overwrites matching keys; it does not add numerical values together.

````python
stock = {"pens": 12, "books": 5}
stock["pencils"] = 8
stock["pens"] = stock["pens"] + 3
print("After assignment:", stock)
result = stock.update({"books": 7, "erasers": 4})
print("After update:", stock)
print("Return value:", result)
````

### Choose a removal operation

| Operation | Behaviour |
|---|---|
| `del d[key]` | Removes a known key; missing key raises `KeyError` |
| `d.pop(key, default)` | Removes and returns its value; returns default if missing |
| `d.popitem()` | Removes and returns the last inserted pair; empty dictionary raises `KeyError` |
| `d.clear()` | Removes all pairs from the existing dictionary |

`update()` and `clear()` change the dictionary and return `None`. Do not assign their result back to your dictionary.

````python
queue = {"T01": "Anu", "T02": "Ravi", "T03": "Sara"}
removed = queue.pop("T02")
print("Removed value:", removed)
print("Missing ticket:", queue.pop("T99", "Not found"))
last_pair = queue.popitem()
print("Last inserted pair:", last_pair)
del queue["T01"]
print("After deletions:", queue)
queue.clear()
print("After clear:", queue)
````

### Default insertion: `setdefault()`

`get()` retrieves without inserting. `setdefault(key, default)` inserts the default only if the key is missing, then returns the stored value. It does not replace an existing value.

````python
visits = {"Anu": 2}
print(visits.setdefault("Anu", 0))
print(visits.setdefault("Ravi", 0))
print(visits)
````

### Insertion order is not alphabetical order

Dictionaries preserve insertion order. Updating an existing key keeps its position. Deleting and reinserting a key puts it at the end. Dictionaries do not support positional slicing. Dictionary equality compares pairs, not insertion order.

````python
labels = {"B": 2, "A": 1}
labels["B"] = 20
print("Updated B:", list(labels))
labels.pop("B")
labels["B"] = 20
print("Reinserted B:", list(labels))
print({"A": 1, "B": 20} == {"B": 20, "A": 1})
````

# Stage 4 — Traverse and summarise · 30 minutes

Loop directly over a dictionary for its keys. Use `values()` for values and `items()` for key–value pairs. In `for key, value in d.items()`, Python unpacks each two-item pair.

````python
marks = {"Anu": 78, "Ravi": 65, "Sara": 91}
print("Keys:")
for name in marks:
    print(name)
print("Values:", list(marks.values()))
print("Pairs:")
for name, mark in marks.items():
    print(name, "scored", mark)
print("Mean:", round(sum(marks.values()) / len(marks), 2))
````

### Views and snapshots

`keys()`, `values()`, and `items()` return views that reflect later changes. `list(d.keys())` creates a snapshot of the keys at that moment.

````python
prices = {"tea": 15, "coffee": 25}
key_view = prices.keys()
key_snapshot = list(prices.keys())
prices["juice"] = 40
print("Live view:", key_view)
print("Earlier snapshot:", key_snapshot)
````

### Be careful with empty dictionaries and deletion

An empty dictionary has no mean or maximum. Check it before dividing by its length or calling `max()`.

Adding or deleting keys while iterating directly over a dictionary is unsafe and can raise `RuntimeError`. Collect keys first, or iterate over a snapshot.

````python
readings = {}
if readings:
    print(sum(readings.values()) / len(readings))
else:
    print("No readings available.")

balances = {"A": 0, "B": 10, "C": 0}
to_remove = []
for account, balance in balances.items():
    if balance == 0:
        to_remove.append(account)
for account in to_remove:
    del balances[account]
print(balances)
````

### Produce a sorted report

`sorted(d)` produces a list of sorted **keys**; it does not sort by values or mutate the dictionary. A dictionary has no `sort()` method.

````python
scores = {"Ravi": 65, "Anu": 78, "Sara": 91}
for name in sorted(scores):
    print(name, scores[name])
print("Original key order:", list(scores))
````

# Stage 5 — Count and group · 30 minutes

A counting dictionary stores `item → number of occurrences`. For every input item, retrieve its old count (or zero) and add one.

````python
responses = ["yes", "no", "yes", "maybe", "no", "yes"]
counts = {}
for response in responses:
    counts[response] = counts.get(response, 0) + 1
print(counts)
````

### Grouping stores lists as values

Counting stores one number per key. Grouping stores all matching items together. Create a separate list when each new group first appears.

````python
registrations = [
    ["Python", "Anu"],
    ["Statistics", "Ravi"],
    ["Python", "Sara"]
]
groups = {}
for subject, name in registrations:
    if subject not in groups:
        groups[subject] = []
    groups[subject].append(name)
print(groups)
# Equivalent insertion pattern, for later practice:
# groups.setdefault(subject, []).append(name)
````

# Stage 6 — Nested dictionaries and copying · 25 minutes

A nested dictionary maps an identifier to a record. Each record is itself a dictionary. Lists inside a record can store repeated observations.

````python
students = {
    "MDS001": {"name": "Anu", "marks": [70, 80]},
    "MDS002": {"name": "Ravi", "marks": [65, 75]}
}
print(students["MDS001"]["name"])
students["MDS001"]["marks"].append(90)
for register, record in students.items():
    print(register, record["name"], record["marks"])
````

### Assignment, shallow copy, and nested changes

`alias = original` gives another name to the same dictionary. `original.copy()` creates a new outer dictionary, but nested mutable objects are still shared. Replacing an outer value differs from changing a shared inner list.

````python
original = {"name": "Anu", "marks": [70, 80]}
alias = original
shallow = original.copy()
shallow["name"] = "Sara"
shallow["marks"].append(90)
print("Original:", original)
print("Alias:", alias)
print("Shallow copy:", shallow)
print("Alias is original:", alias is original)
````

To copy this particular two-field record independently, copy its marks list too. For more deeply nested general data, the standard library offers `copy.deepcopy()`; it is an optional topic.

````python
record = {"name": "Anu", "marks": [70, 80]}
independent = {"name": record["name"], "marks": record["marks"].copy()}
independent["marks"].append(95)
print("Source:", record)
print("Independent:", independent)
````

### A shared-default trap: `fromkeys()`

`dict.fromkeys(keys, value)` assigns the **same value object** to every key. This is convenient for an immutable default such as zero, but sharing a list may be unintended.

````python
print(dict.fromkeys(["Mon", "Tue"], 0))
shared = dict.fromkeys(["Anu", "Ravi"], [])
shared["Anu"].append("Python")
print("Shared-list result:", shared)

separate = {}
for name in ["Anu", "Ravi"]:
    separate[name] = []
separate["Anu"].append("Python")
print("Separate-list result:", separate)
````

## Optional syntax preview — dictionary comprehensions

A dictionary comprehension builds a dictionary from an expression and a loop. Learn the ordinary loop first; both forms below calculate the same result.

````python
quantities = {"pens": 12, "books": 3, "erasers": 0}
low_stock_loop = {}
for item, quantity in quantities.items():
    if quantity < 5:
        low_stock_loop[item] = quantity
low_stock_comp = {item: quantity for item, quantity in quantities.items() if quantity < 5}
print(low_stock_loop)
print(low_stock_comp)
````

## Quick reference

| Need | Syntax | Important detail |
|---|---|---|
| Empty dictionary | `d = {}` | Empty mapping |
| Retrieve known key | `d[key]` | Missing key raises KeyError |
| Retrieve safely | `d.get(key, default)` | Does not insert |
| Add or replace | `d[key] = value` | Keys remain unique |
| Several changes | `d.update(other)` | Matching keys overwritten; returns None |
| Insert default | `d.setdefault(key, default)` | Returns stored value |
| Check key | `key in d` | Checks keys |
| Keys / values / pairs | `d.keys()`, `d.values()`, `d.items()` | Dynamic views |
| Remove known key | `del d[key]` | Missing key raises KeyError |
| Remove and retrieve | `d.pop(key, default)` | Missing key uses default |
| Remove last inserted pair | `d.popitem()` | Empty dictionary raises KeyError |
| Empty existing dictionary | `d.clear()` | Returns None |
| Shallow copy | `d.copy()` | Nested mutable values still shared |
| Initialise same default | `dict.fromkeys(keys, value)` | One shared value object |
| Count pairs | `len(d)` | Counts keys, not nested elements |
| Ordered report keys | `sorted(d)` | Returns a new list of keys |

**Useful patterns:** `counts[x] = counts.get(x, 0) + 1`; `for key, value in d.items():`; check for empty data before calculating a mean or maximum.
